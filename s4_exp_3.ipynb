# **EXPERIMENT 3**

Calculate the measures of dispersion, namely range and standard deviation for a given
data.

###**Aim**

To calculate and display the range and standard deviation of a given dataset. These are
measures of dispersion that quantify the spread or variability of the data points.

###**Objective**
1. Range provides a basic understanding of the spread of the data.
2. The standard deviation gives a more nuanced view of data variability.

###**Algorithm**

Step 1: Import the statistics module.

Step 2: Define the dataset.

Step 3: Calculate the Range and standard deviation

Step 4: Output the results.

**Case**: Find the range and standard deviation for a given data 10, 15, 8, 22, 17, 13, 9.

```python
import statistics
# Given data
data = [10, 15, 8, 22, 17, 13, 9]
# Calculate Range
data_range = max(data) - min(data)
# Calculate Standard Deviation
std_dev = statistics.stdev(data)
# Output
print("Data:", data)
print("Range:", data_range)
print("Standard Deviation:", round(std_dev, 2))
```

### **Result and Discussion**

The measures of dispertion provide insights into the spread and variability within the given dataset. A higher standard deviatrion suggests a wider spread of data points, while a lower standard deviation of 4.96 suggests moderate variability within the dataset.

##**Application Problem: Analysis of manufacturing tolerances**

**Problem Statement**

A quality engineer in a manufacturing plant is analyzing the diameter (in mm) of metal
rods produced by a CNC machine and are given by
10.01, 10.03, 9.98, 10.05, 10.00,
10.02, 10.04, 9.99, 10.01, 10.03,9.97, 10.06, 10.02, 10.01, 9.98, 10.00, 10.04, 9.96, 10.02,
10.01(rod diameter in mm).
Consistency is critical—too much variation might lead to assembly failures. To assess precision, the engineer calculates range and standard
deviation of the rod diameters.


```python
import numpy as np
diameters = [10.01, 10.03, 9.98, 10.05, 10.00,10.02,10.04,9.99,10.01,10.03,9.97,10.06,10.02,10.01,9.98,10.00,10.04,9.96,10.02,10.01]
#rod diameter in mm
range_diameter = max(diameters) - min(diameters)
std_diameter = np.std(diameters) # population standard deviation
print(f"Range of Diameters: {range_diameter:.4f} mm")
print(f"Standard Deviation of Diameters: {std_diameter:.4f} mm")
#Visualize the dispersion
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10, 5))
sns.histplot(diameters, bins=10, kde=True, color='lightcoral', edgecolor='black')
plt.axvline(np.mean(diameters), color='blue', linestyle='--', label=f'Mean: {np.mean(diameters):.4f}')
plt.axvline(np.mean(diameters) + std_diameter, color='green', linestyle=':', label ='+1 Std Dev')
plt.axvline(np.mean(diameters) - std_diameter, color='green', linestyle=':', label ='-1 Std Dev')

plt.title("Distribution of Rod Diameters")
plt.xlabel("Diameter (mm)")
plt.ylabel("Frequency")
plt.legend()
plt.grid(True)
plt.show()
#Range shows the full spread of diameters.
#Standard deviation reveals how tightly the diameters cluster around the mean.
#A small std. dev. and tight range mean high precision in manufacturing.
```

## **Application Problem: Analysis of manufacturing tolerance**

**Problem Statement**

An industrial engineering team is analyzing the temperature readings from a sensor
installed in a high-precision manufacturing machine. The sensor records the
temperature every hour. The following temperature readings (in °C) were recorded over a
10-hour shift are 22.4, 22.7, 22.5, 22.9, 22.8, 22.6, 22.7, 22.3, 22.9, 22.6.Calculate the Range
and standard deviation of the temperature readings.

```python
import numpy as np
#Temperature readings
data =  [22.4, 22.7, 22.5, 22.9, 22.8, 22.6, 22.7, 22.3, 22.9, 22.6]
# Calculate range
range_value = max(data)-min(data)
# Calculate Standard deviation
std_value = np.std(data, ddof=0) # Use ddof=1 for sample std dev
print(f"Range: {range_value:.2f} °C")
print(f"Standard Deviation: {std_value:.2f} °C")
```
